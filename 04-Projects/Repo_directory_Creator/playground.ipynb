# Automating python directory creation v1

### creating folder directory using below code block

- 📁 project_name 
- 📁 src
-   ∟ 📄.env
-   ∟ 📄README.md
-   ∟ 📄requirements.txt
-   ∟ 📄main.py

```python
from pathlib import Path

project = Path(input("Enter project name: "))

# 1. Create folders only if they don't exist
if not project.exists():
    project.mkdir()
    print(f"📁 {project} created!")
if not (project / "src").exists():
    (project / "src").mkdir()
    print("📁 src created!")

# 2. Create the empty files
for file in [".env", "README.md", "requirements.txt", "main.py"]:
    (project / file).touch()
    print(f"✅ Project '{file}' created successfully!")
```
