## 1. Below code won't work due to anti-bot protection Enabled to the website w3schools.com ❌
- You don't have permission to access ...

```python
import requests

# the required first parameter of the 'get' method is the 'url':
url = "https://w3schools.com/python/demopage.htm"
r = requests.get(url)

# print the response text (the content of the requested file):
print(r.status_code)
print(r.text) # anti-bot protection → You don't have permission to access ...
```

# 2. Standard Best Approach: ✅

Add a browser-like User-Agent

```python
import requests

url = "https://www.w3schools.com/python/demopage.htm"

headers = {
    "User-Agent": "Mozilla/5.0"
}

r = requests.get(url, headers=headers)

print(r.status_code)
print(r.text)
```

### Websites Which Don't Have Anti-bot Protection

```python
import requests

r = requests.get("https://httpbin.org/get")
print(r.status_code)
print(r.text)
```
