````python
# explanation: suffix returns only the last file extension, while suffixes returns all file extensions.

p = Path("archive.tar.gz")
print(p.suffix) # gz

# further more code for practice executed
print(p.suffixes)  # ['.tar', '.gz']
print(p.stem)  # archive.tar
print(p.name)  # archive.tar.gz
````

**Q20.** Write code that prints `"Found it!"` only if a file called `"data.csv"` exists - and prints `"Not found"` otherwise.

````python
# Write your code here
# Path("data.csv").mkdir()
# Path("data.csv").rmdir()

if not Path("data.csv").exists():
    Path("data.csv").touch()
    print("Found it!")
elif Path("data.csv").exists():
    print("Found it and deleted!")
    Path("data.csv").unlink()
    print("Not found!")
````

**Q21.** Why would you check `.exists()` before calling `.mkdir()` if you were NOT using `exist_ok=True`? Answer in one sentence.

````python
# answer: To avoid FileExistsError if the directory already exists.
from pathlib import Path

# --- check name already exist ---
# print(Path("sample_dir").exists())  # True

# --- return FileExistsError if name already exists ---
# print(Path("sample_dir").mkdir())   # FileExistsError


# --- create dir or inform dir already exists ---
path = Path("sample_dir")

try:
    path.mkdir()  # using mkdir(exist_ok=True) FileExistsError never trigger
    print(f"{path} created")
except FileExistsError:
    print(f"{path} already exist")
````
