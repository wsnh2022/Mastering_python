### The Core Concept

`pathlib` treats file paths as objects rather than simple strings. This means you don't have to worry about operating system differences (like Windows using `\` and Mac/Linux using `/`)—`pathlib` handles the translation automatically.

**Key Mechanics:**

1. **Creation:** `Path('filename.txt')`, `Path.cwd()` (current working directory), or `Path.home()` (user's home directory).
2. **Joining:** Use the forward slash `/` operator to combine paths instead of string concatenation. Example: `Path.cwd() / 'Documents' / 'notes.txt'`.
3. **Properties (no parentheses):** Extract parts of the path using `.name` (file.txt), `.stem` (file), `.suffix` (.txt), and `.parent` (the folder containing it).
4. **Checking State (with parentheses):** Verify what's actually on your drive using `.exists()`, `.is_file()`, and `.is_dir()`.

### Level 2: Intermediate

**Question 1:**
You have the following list of strings:
`files = ['data.csv', 'script.py', 'config.csv', 'notes.txt']`

Write a loop that iterates through this list. For each string:

1. Convert it to a `Path` object.
2. If the file is a `.csv` file, print its `.stem` (just the name).
3. If it is not a `.csv` file, ignore it (do nothing).

Show me your code.

````python
from pathlib import Path

files = ['data.csv','script.py','config.csv', 'notes.txt']

for file in files:
    p = Path(file)
    if p.suffix == '.csv':
        print(p.stem)
    elif p.suffix != ".csv":
        continue
````

That is correct. It successfully identifies the `.csv` files and extracts just the stem.

*A quick code review note on readability:* The `elif p.suffix != ".csv": continue` block isn't strictly necessary here. If an `if` condition isn't met in a `for` loop, Python automatically moves on to the next item anyway. You can just drop the `elif` completely.

**Question 2:**
You want to save a configuration string `"debug=True"` to a file named `settings.ini` in the current working directory, but you must ensure you don't overwrite an existing file.

Write a snippet that:

1. Creates the `Path` object for `settings.ini` inside the current working directory.
2. Checks if the file already exists.
3. If it does *not* exist, write the text `"debug=True"` into the file.
4. If it *does* exist, print `"File already exists."`

Show me your code.

````python
from pathlib import Path

file = Path("settings.ini")

if file.exists():
    print("File already exists.")
else:
    file.touch()
    file.write_text("debug=True")
````

That is correct. It successfully guards against overwriting an existing file.

*Code review note:* `write_text()` actually creates the file automatically if it doesn't exist. Calling `touch()` right before it works perfectly fine, but is slightly redundant. You can just call `write_text()` directly in the `else` block.

**Question 1:**
You need to write a log message to a file, but the folder for the logs might not exist yet. Attempting to write a file into a non-existent folder will crash your program.

Write a snippet that:

1. Creates a `Path` object for a folder named `logs` in the current working directory.
2. Creates a `Path` object for a file named `app.log` inside that `logs` folder.
3. Ensures the `logs` folder actually exists on the drive (create it if it doesn't, but make sure your code doesn't crash if the folder is already there).
4. Writes the string `"System started"` into the `app.log` file.

Show me your code.

````python
from pathlib import Path

folder = Path("logs") / "apps.log"
folder.parent.mkdir(parents=True, exist_ok=True)
folder.touch(exist_ok=True)
folder.write_text("System started")  #System started → char count 14
````

That is a clean, perfectly executed solution.

By targeting `folder.parent` and applying `parents=True` along with `exist_ok=True` on `mkdir()`, you correctly handled both edge cases—creating missing parent directories and preventing crashes if the folder already exists.

**Question 1:**
You are building a backup utility. The system provides you with a source file path as a string:
`source = 'data/user_exports/2026_report.csv'`

Your program needs to construct a target destination path inside a `backup` folder. The final path needs to meet three business rules:

1. It must preserve the exact file name and extension (`2026_report.csv`).
2. It must add `.bak` as a secondary extension to the end, resulting in `2026_report.csv.bak`.
3. It must place this file inside a `backup` folder relative to the current working directory.

Construct this target `Path` object using `pathlib` methods/properties rather than plain string manipulation or string formatting.

Show me your code.

````python
from pathlib import Path

source = "data/user_exports/2026_report.csv"
source_path = Path(source)
# print(source_path)
# print(source_path.suffix)

target_path = Path("backup") / source_path.with_suffix(source_path.suffix + ".bak").name

print(target_path)
````

That is a clever design choice and a completely valid use of `with_suffix()` and `.name` to construct the target path.
