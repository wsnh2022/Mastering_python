## The 5 rules worth memorizing

Never commit .venv/.

Never commit .env secrets.

Never commit __pycache__/.

Ignore editor and OS junk files.

Commit source code, notebooks, notes, and project configuration files.

# The 5 Git rules every Python beginner should memorize

## 1. Never commit `.venv/`

### What is it?

Your virtual environment - a local copy of installed packages.

```text
.venv/
├── Scripts/
├── Lib/
└── ...
```

### Why ignore it?

* Can be hundreds of MB.
* Different on Windows, macOS, and Linux.
* Can be recreated anytime.

### Professional workflow

Commit **requirements.txt** or **pyproject.toml**, not the environment itself.

```bash
pip freeze > requirements.txt
```

Another developer runs:

```bash
python -m venv .venv
pip install -r requirements.txt
```

### Memory trick

**Commit the recipe, not the cooked meal.**

---

## 2. Never commit `.env` secrets

### What is it?

Environment variables such as API keys and passwords.

```env
OPENROUTER_API_KEY=sk-xxxx
DB_PASSWORD=secret123
```

### Why ignore it?

If pushed to GitHub, the secret is effectively exposed.

### Best practice

Keep a safe template:

```text
.env.example
```

Example:

```env
OPENROUTER_API_KEY=
DB_PASSWORD=
```

Developers copy it locally to `.env`.

### Memory trick

**Secrets stay on your machine, not in Git.**

---

## 3. Never commit `__pycache__/`

### What is it?

Compiled Python bytecode files.

```text
__pycache__/
main.cpython-313.pyc
```

Python generates these automatically when code runs.

### Why ignore it?

* Not source code.
* Changes constantly.
* Different Python versions create different files.

### Memory trick

**Python can regenerate it in seconds.**

---

## 4. Ignore editor and OS junk files

### Examples

```text
.vscode/
.idea/
.DS_Store
Thumbs.db
desktop.ini
```

### Why ignore them?
They contain personal editor settings or operating-system metadata.

### Exception
Teams sometimes keep selected VS Code files:

```gitignore
.vscode/*
!.vscode/settings.json
```

### Memory trick

**Ignore anything that describes your computer instead of the project.**

---

## 5. Commit source code, notebooks, notes, and configuration

These are the files that matter.

### Commit these

```text
main.py
app.py
README.md
notes.md
analysis.ipynb
requirements.txt
pyproject.toml
.gitignore
```

These files contain knowledge, logic, or instructions.

### Ask before committing

**“If I delete my laptop tomorrow, would I want this file from GitHub?”**

* Yes → commit it.
* No → ignore it.

---

# The easiest mental model

| File type          | Commit?    | Why                |
| ------------------ | ---------- | ------------------ |
| `.py`              | Yes        | Your code          |
| `.ipynb`           | Yes        | Your learning/work |
| `.md`              | Yes        | Documentation      |
| `requirements.txt` | Yes        | Dependency recipe  |
| `.venv/`           | No         | Rebuildable        |
| `__pycache__/`     | No         | Rebuildable        |
| `.env`             | No         | Secret             |
| `.DS_Store`        | No         | OS junk            |
| `.vscode/`         | Usually no | Personal settings  |

---

# The 10-second Git test

Before every commit, quickly classify files:

### Keep

* Code
* Notes
* Notebooks
* Config
* README

### Ignore

* Installed packages
* Cache
* Build output
* Logs
* Secrets
* Editor junk

If a file falls into the second list, add it to `.gitignore`.

# One sentence to remember forever

> **Git stores your work; `.gitignore` excludes everything that can be recreated, is personal, or is secret.**


# Default Git Ignore Template

```gitignore
# ----------------
# Python
# ----------------
__pycache__/
*.py[cod]
*$py.class

# Virtual environments
.venv/
venv/
env/
ENV/

# ----------------
# Jupyter Notebook
# ----------------
.ipynb_checkpoints/

# ----------------
# Test / Lint / Type-check caches
# ----------------
.pytest_cache/
.mypy_cache/
.ruff_cache/
.coverage
htmlcov/

# ----------------
# Build / Packaging
# ----------------
build/
dist/
*.egg-info/
.eggs/
pip-wheel-metadata/

# ----------------
# IDE / Editor
# ----------------
.vscode/*
!.vscode/settings.json
!.vscode/extensions.json
!.vscode/launch.json

.idea/

# Workspace files
*.code-workspace

# ----------------
# Operating system files
# ----------------
.DS_Store
Thumbs.db
desktop.ini

# ----------------
# Logs
# ----------------
*.log

# ----------------
# Environment variables / secrets
# ----------------
.env
.env.*
!.env.example

# ----------------
# SQLite databases (optional)
# Uncomment if databases should not be committed
# ----------------
# *.db
# *.sqlite3

# ----------------
# Playwright / Browser automation
# ----------------
playwright-report/
test-results/

# ----------------
# Temporary files
# ----------------
tmp/
temp/
*.tmp
```
