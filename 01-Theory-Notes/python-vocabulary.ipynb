# 1. python_dev_vocabulary (workplace version)

## Golden rule

**Installed with `pip` → call it a *package***

* `pip install requests` → **requests package**
* `pip install pandas` → **pandas package**
* `pip install playwright` → **playwright package**

---

## Built into Python → call it a *module*

* `import os` → **os module**
* `import json` → **json module**
* `import pathlib` → **pathlib module**

---

## Bigger architecture tool → call it a *framework*

* **FastAPI framework**
* **Django framework**
* **Airflow framework** (or orchestration framework)

---

## What to say in meetings

### Say this

* “I’m using the **requests package**.”
* “This package is already in our dependencies.”
* “It’s part of the **Python standard library**.”
* “We can add the package to **requirements.txt**.”
* “Playwright is our browser automation package.”

### Avoid this

* “requests library package module”
* “pip library”
* “Python software package library”

---

## One-line memory trick

* **pip = package**
* **import from Python = module**
* **full application structure = framework**

---

## Interview / boss-ready sentence

> “I’m using the `requests` package for HTTP calls and the `playwright` package for browser automation; `json` comes from the Python standard library.”

## 11. Communication phrases that sound professional

### Instead of

* “It is not working.”

### Say

* “The job is failing.”

* “I can reproduce the issue.”

* “The API is returning a 403.”

* “I’m investigating the root cause.”

* “I’ve identified the bottleneck.”

* “The fix is ready for review.”

* “I’ll deploy after validation.”
