# Print Without a New Line

### **Explanation:**
- `sep="-"` inserts a hyphen between multiple arguments.
- `end="!\n"` replaces the default newline with `!` followed by a newline.


````python
print('i am', 35, 'years old.') # combining numbers and text in 1 output by seperating with comma
print("hai", "world", end="!")  # end replaces default newline with "!"
print("hai", "world", end="\n") # end explicitly set to newline (default behavior)

# sep replaces default space with "-"
print("hello world")
````

# Print Numbers

````python
# Numbers Arthmetic Solving Order is followed by this rule → PEMDAS
# paranthesis    →  ()
# Exponent       →  **
# Multiplication →   *
# division       →   /
# addition       →   +
# Subtractions   →   -

# PEMDAS solving order
print((3 ** 3) + 3 - 10 * 2 / 2 ) # 20.0

# controlled arthematic possible using Paranthesis
print(((3 ** 4) + 4) - (10 * (10 / 2))) # 35.0

# Step-by-step process:
#   3 ** 4 = 81
#   81 + 4 = 85
#   10 / 2 = 5.0
#   10 * 5.0 = 50.0
#   85 - 50.0 = 35.0
````
