### 3 Easier Variations (single code block)

---

````python
# Variation 1
# New requirement: Split comma-separated names and remove extra spaces.
# Core idea stays the same: split text into items and clean spaces.
# New concept added: Different separator (,).

text = "Alice, Bob , Charlie ,  David"
split_txt = text.split(",")
clean = [item.strip() for item in split_txt]
print(clean)

# Variation 2
# New requirement: Split semicolon-separated fruits and clean each item.
# Core idea stays the same: split and strip items.
# New concept added: Semicolon separator (;).

fruits = "apple ; banana; mango ;grapes"
fruits1 = fruits.split(";")
clean1 = [item.strip() for item in fruits1]
print(clean1)

# Variation 3
# New requirement: Split a sentence into words and clean accidental spaces.
# Core idea stays the same: create a cleaned list from text.
# New concept added: Splitting by whitespace.

sentence = "  learn   python   automation  today  "
sentence1 = sentence.split()            # any amount of whitespace (spaces, tabs, newlines) as a separator.
clean2 = [item.strip() for item in sentence1]
print(clean2)
````

---

### Learned Notes:

#### when `text.split()` treats one or more whitespace characters (space, tab, newline, or a mix of them) as a single separator and removes extra whitespace automatically.

```python
print("learn python automation".split())              # single spaces
print("learn   python     automation".split())        # multiple spaces
print("   learn python automation   ".split())        # leading/trailing spaces
print("learn\tpython\tautomation".split())            # tabs
print("learn\npython\nautomation".split())            # newlines
print("learn \t python\n\nautomation".split())        # mixed whitespace
```

### same output for all 
### `['learn', 'python', 'automation']`

## 3 Similar-Level Variations (single code block)

````python
# Variation 1
# New requirement: Split pipe-separated subjects and convert each item to uppercase.
# Core idea stays the same: split and clean items.
# New concept added: String transformation with upper().

subjects = " math | science | english | history "
sub = subjects.split("|")
uppercase = [item.strip().upper() for item in sub]
print(uppercase)


# Variation 2
# New requirement: Split comma-separated cities and convert each item to title case.
# Core idea stays the same: clean each split item.
# New concept added: title() formatting.

cities = "chennai, hyderabad , bengaluru , mumbai"
cities = cities.split(",")
cities = [item.strip().title() for item in cities]
print(cities) # ['Chennai', 'Hyderabad', 'Bengaluru', 'Mumbai']

# Variation 3
# New requirement: Split tags and remove empty items caused by repeated separators.
# Core idea stays the same: split and clean items.
# New concept added: Filtering unwanted values.

tags = "python, automation, , ai, , data "

parts = tags.split(",")
cleaned = [i.strip() for i in parts if i.strip()]

print(cleaned)
````

````python
parts = ["python", "   ", " ai ", "", " data "]
stripped = [i.strip() for i in parts]
print(stripped)

stripped = [i.strip() for i in parts if i.strip()]
print(stripped)
````
