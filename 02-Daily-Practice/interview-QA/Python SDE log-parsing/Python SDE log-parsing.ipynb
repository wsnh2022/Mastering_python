# Step 13 - Count ERROR lines

Combine loop + split + if + dictionary.


```python
logs = [
    "t | payment-service | ERROR | x",
    "t | payment-service | INFO | y",
    "t | payment-service | ERROR | z"
]

counts = {}

for line in logs:
    parts = [p.strip() for p in line.split("|")]
    service = parts[1]
    level = parts[2]

    if level == "ERROR":
        if service not in counts:
            counts[service] = 0

        counts[service] += 1

print(counts)
```

# Step 15 - Function

Create a reusable function.


```python
def error_counts(logs):
    counts = {}

    for line in logs:
        parts = [p.strip() for p in line.split("|")]

        if len(parts) < 4:
            continue

        service = parts[1]
        level = parts[2]

        if level == "ERROR":
            counts[service] = counts.get(service, 0) + 1

    return counts


logs = [
    "t | payment-service | ERROR | x",
    "t | auth-service | ERROR | y"
]

print(error_counts(logs))
```

# Step 19 - Final function

Build top_errors().


```python
def error_counts(logs):
    counts = {}

    for line in logs:
        parts = [p.strip() for p in line.split("|")]

        if len(parts) < 4:
            continue

        service = parts[1]
        level = parts[2]

        if level == "ERROR":
            counts[service] = counts.get(service, 0) + 1

    return counts


def top_errors(logs, n):
    counts = error_counts(logs)

    items = list(counts.items())
    items.sort(key=lambda x: (-x[1], x[0]))

    return items[:n]


logs = [
    "t | payment-service | ERROR | x",
    "t | payment-service | ERROR | y",
    "t | auth-service | ERROR | z"
]

print(top_errors(logs, 2))
```
